```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def convolution(k,A):
    #padding required
    shape_A=A.shape
    shape_k=k.shape
    a= np.zeros((shape_A[0]+2*(shape_k[0]-1),shape_A[1]+2*(shape_k[1]-1)))
    a[shape_k[0]-1:shape_k[0]-1+shape_A[0],shape_k[1]-1:shape_k[1]-1+shape_A[1]]=A
    shape_a=a.shape

    # stride
    stride=1
    shape_b= ((shape_a[0]-shape_k[0])//stride +1 ,(shape_a[1]-shape_k[1])//stride +1)

    #b is output
    b= np.zeros(shape_b)
    
    #convolution
    row=0
    for i in range(0,shape_a[0]-shape_k[0]+1,stride):
        col=0
        for j in range(0,shape_a[1]-shape_k[1]+1,stride):
            b[row][col]=np.sum(k*a[i:i+shape_k[0],j:j+shape_k[1]])
            col+=1
        row+=1

    return b

    
```

```python
def maxpool(A,shape):
    shape_a=A.shape
    stride= 2
    shape_b= ((shape_a[0]-shape[0])//stride +1 ,(shape_a[1]-shape[1])//stride +1)

    #b is output
    b= np.zeros(shape_b)
    row=0
    for i in range(0,shape_a[0]-shape[0]+1,stride):
        col=0
        for j in range(0,shape_a[1]-shape[1]+1,stride):
            b[row][col]=np.max(A[i:i+shape[0],j:j+shape[1]])
            col+=1
        row+=1

    return b
```
